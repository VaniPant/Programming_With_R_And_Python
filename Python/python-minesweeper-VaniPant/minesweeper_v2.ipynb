# MineSweeper 💣⛳️

We will work around the classic game of MineSweeper. The game consists of a grid of squares, some of which contain mines. If you click on a square containing a mine, you lose. If you click on a square that doesn't contain a mine, you reveal the number of mines in the adjacent squares. The game is won when all squares without mines are revealed. For more info check the [Wikipedia page](https://en.wikipedia.org/wiki/Minesweeper_(video_game)).

## Place Mines

Now here we will implement a function to place mines in the grid. The function will take the grid and the number of mines to place. It will randomly place the mines in the grid.

```python
import random

def put_bombs(grid: list[list[dict]], bomb_count: int):
    """
    Randomly place mines in the grid.

    Parameters:
    grid (list): A 2D list representing the grid, where each element is a dictionary representing a cell.
    bomb_count (int): The number of mines to place in the grid.
    """
    n = len(grid)
    m = len(grid[0])
    placed_bombs = 0
    while placed_bombs < bomb_count:
        i = random.randint(0, n - 1)
        j = random.randint(0, m - 1)
        if not grid[i][j]["mine"]:
            grid[i][j]["mine"] = True
            placed_bombs += 1
            for x in range(max(0, i - 1), min(n, i + 2)):
                for y in range(max(0, j - 1), min(m, j + 2)):
                    if not grid[x][y]["mine"]:
                        grid[x][y]["adjacent_mines"] += 1
    return grid


# ✏️ We can see some mines can overlap. Fix this bug [4 point]
```

# Let's play 🎮

Now we will implement a function that performs users actions on the grid.
If player clicks on a cell with a zero adjacent mines, we will reveal all the adjacent cells untill we reach cells with adjacent mines.
 

```python
def move(grid: list[list[dict]], row: int, col: int, click: bool):
    """
    Make a move in the grid.

    Parameters:
    grid (list): A 2D list representing the grid, where each element is a dictionary representing a cell.
    row (int): The row of the cell to click.
    col (int): The column of the cell to click.
    click (bool): Whether to click or flag the cell. True for click, False for flag.

    Returns:
    bool: Whether the move resulted in a mine explosion.
    """
    cell = grid[row][col]
    if click:
        if cell["flagged"]:
            # Don't reveal flagged cells
            return False
        if cell["mine"]:
            # Mine explosion return one bool and the updated grid
            # ✏️ Your code here [0.5 point]
            # pass
            print("Boom! You hit a mine!")
            return True
        cell["revealed"] = True
        if cell["adjacent_mines"] == 0:
            # If the cell has no adjacent mines, reveal all adjacent cells
            # if you clicked on a cell with no adjacent mines, you should reveal all adjacent cells
            # ✏️ Your code here [2 points]
            # pass
            for x in range(max(0, row - 1), min(len(grid), row + 2)):
                for y in range(max(0, col - 1), min(len(grid[0]), col + 2)):
                    if not grid[x][y]["revealed"]:
                        move(grid, x, y, True)
    else:
        # Toggle the flagged status of the cell
        cell["flagged"] = not cell["flagged"]

    # ✏️ Describe why we dont need to write: grid[row][col] = cell [0.5 point]
    # While assigning cell = grid[row][col], the cell does not create a new copy; it simply points to the same dictionary object in the grid. Therefore, any changes made to the cell are directly reflected in the grid. As a result, there's no need to reassign the cell back to grid[row][col] because the grid is automatically updated.
    # ✏️ Describe why we dont need to return the grid [0.5 point]
    # The grid variable is a list of lists containing dictionaries. By passing the grid into the move function and modifying it within the function, we can directly change the original grid that exists outside the function. This is because we are working with the same grid object in memory. Therefore, there is no need to return the grid from the function, as the changes are already applied to the original grid.
    return False
```

To approach this, a straightforward method involves generating all possible grid configurations and validating them against the revealed cells. Each cell can either contain a bomb or not, giving us $2^{n \cdot m}$ possible grids for an $n \times m$ grid. However, this approach becomes computationally infeasible for large grids.

To start, let's consider that the structure of the grid that we were using, should not be used anymore. We need to create a new structure that does not have the information about the unrevealed cells. To warm up, let's implement a function that will return the grid without the unrevealed cells.

```python
anonymized_grid_type = list[list[int]]


def anonymized_grid(grid: list[list[dict]]) -> anonymized_grid_type:
    """
    Anonymize the grid by removing the mine and adjacent_mines information.
    unrevealed cells should be -1, other cells should be in the range [0, 8].

    Parameters:
    grid (list): A 2D list representing the grid, where each element is a dictionary representing a cell.

    Returns:
    list: A 2D list representing the grid, where each element is an integer representing a cell.
    """
    anonym_grid = [
        [-1 if not cell["revealed"] else cell["adjacent_mines"] for cell in row]
        for row in grid
    ]
    return anonym_grid


def show_anomymized_grid(grid: anonymized_grid_type):
    """
    Display the anonymized grid.

    Parameters:
    grid (list): A 2D list representing the grid, where each element is an integer representing a cell.
    """
    start_end_line = " + " + " ".join(str(i) for i in range(len(grid[0]))) + " +"
    print(start_end_line)
    for i, row in enumerate(grid):
        print(f"{i}|", end=" ")
        for cell in row:
            match cell:
                case -1:
                    val = UNREVEALED
                case 0:
                    val = " "
                case _:
                    val = str(cell)
            print(val, end=" ")
        print("|")
    print(start_end_line)


random.seed(21)  # to make the game deterministic
grid = create_plane_grid(10, 10)
grid = put_bombs(grid, 10)
move(grid, 0, 0, True)
show_anomymized_grid(anonymized_grid(grid))
```

## The Algorithm

The algorithm closely mimics how humans play the game. It systematically identifies **obvious mines** and **safe cells** and takes appropriate actions. This process repeats until the game is either won or lost.

Here’s how the algorithm works:

### 1. Identifying **Obvious Mines**
   - A mine is considered **obvious** if the number on a revealed cell is equal to the number of adjacent unrevealed cells.  
     In this case, all the adjacent unrevealed cells must be mines, so we flag them.

### 2. Identifying **Safe Cells**
   - A cell is considered **safe to click** if the number of flagged cells around a revealed cell is equal to the number of mines it indicates.  
     In this case, all other adjacent unrevealed cells are safe and can be clicked.

### Steps of the Algorithm:
1. Start by analyzing the revealed cells on the board.
2. Flag all **obvious mines** based on the rules above.
3. Click all **safe cells** identified using the safe-cell rule.
4. Repeat steps 2 and 3 until the game is won or there are no more moves to make (which might require guessing).

By following these rules, the algorithm mimics logical reasoning used in the game, ensuring clear and systematic decision-making.

## Find Neighbors of Revealed Cells

Now we will implement a function that will return the neighbors of revealed cells. The function will take the anonymous grid and return a dictionary with the keys as the coordinates of the revealed cells and the values as the neighbors of the revealed cells.

```python
def find_neighbors(grid: anonymized_grid_type, ignored_values: list[int] = (0, -1)) -> dict:
    """
    Find the unrevealed neighbors of each cell.

    Parameters:
    grid (list): A 2D list representing the grid, where each element is an integer representing a cell.
    ignored_values (list): if a cell has one of these values, it should be ignored. Default is [0, -1].

    Returns:
    dict: A dictionary where the keys are tuples of integers (i, j) representing the coordinates of the cell,
            and the values are lists of tuples, each representing the coordinates of an unrevealed neighboring cell.
    """
    all_neighbors = {}
    for i, row in enumerate(grid):
        for j, cell in enumerate(row):
            if cell in ignored_values:
                # unrevealed cells or cells with no adjacent mines have no neighbors
                continue
            neighbors = []
            # ✏️ iterate over the neighbors of the cell and add the unrevealed neighbors to the `neighbors` list [2 points]
            all_neighbors[(i, j)] = neighbors
    return all_neighbors
```

```python
def find_obvious_mines(grid: anonymized_grid_type):
    """
    Find cells that are obvious mines.

    Parameters:
    grid (list): A 2D list representing the grid, where each element is an integer representing a cell.

    Returns:
    list: A list of tuples, each representing the coordinates of an obvious mine.
    """
    obvious_mines = []
    for (i, j), neighbors in find_neighbors(grid).items():
        if len(neighbors) == grid[i][j]:
            obvious_mines.extend(neighbors)

    # ✏️ Remove duplicates from the `obvious_mines` list (try to do it efficiently, maybe changing the data structure can help) [2 points]
    return obvious_mines
```

Once the obvious mines have been flagged, the next step is to identify the safe cells. To do this:
